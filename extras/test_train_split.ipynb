```python
import os,shutil,random

source_dir = '../../localisation_outputs/train/'
dest_dir = '../../localisation_outputs/test/'

def test_train_split(source_dir, dest_dir, class_name = None):
    class_name += '/'
    source_dir += class_name
    dest_dir += class_name
    no_of_files = round(len(os.listdir(source_dir))*0.2)
    for i in range(no_of_files):
        random_file = random.choice(os.listdir(source_dir))
        source_file = source_dir + random_file
        shutil.move(source_file,dest_dir)
```

```python
for class_name in os.listdir(source_dir):
    test_train_split(source_dir, dest_dir, class_name = class_name)
```

```python
import cv2
import os

# Define the paths for the input and output folders
input_folder = "./MildImpairment/"
output_folder = "./Synthetic_MildImpairment/"

# Define the desired new size for the images
new_size = (128, 128)

for filename in os.listdir(input_folder):
    # Check if the file is an image
    if filename.endswith(".jpg") or filename.endswith(".png"):
        # Load the image in grayscale
        img = cv2.imread(os.path.join(input_folder, filename), cv2.IMREAD_GRAYSCALE)
        # Resize the image
        img = cv2.resize(img, new_size)
        # Save the resized image in the output folder in grayscale
        cv2.imwrite(os.path.join(output_folder, filename), img)

```

```python
import os
import random

folder_path = './last_generator_ModerateImpairment/'
num_images_to_delete = 1440

# Get a list of all files in the folder
all_files = os.listdir(folder_path)

# Filter the list to only include image files
image_files = [f for f in all_files if f.endswith('.jpg') or f.endswith('.png')]

# Choose the number of files to delete
files_to_delete = random.sample(image_files, num_images_to_delete)

# Delete the chosen files
for file_name in files_to_delete:
    file_path = os.path.join(folder_path, file_name)
    os.remove(file_path)
```

```python
import os
import time

# The path to the folder containing the files
src_folder_path = "./shuffled/"   

# The path to the folder where you want to move the files
dst_folder_path = "./Very Mild Impairment/"  

# Get a list of all the files in the folder
files = os.listdir(src_folder_path)

# Generate a random permutation of the indices of the files
permutation = list(range(len(files)))
for i in range(len(permutation)):
    j = int(time.time()) % len(permutation)
    permutation[i], permutation[j] = permutation[j], permutation[i]

# Move the files using the permutation
for i, file in enumerate(files):
    os.rename(os.path.join(src_folder_path, file), os.path.join(dst_folder_path, files[permutation[i]]))
```
